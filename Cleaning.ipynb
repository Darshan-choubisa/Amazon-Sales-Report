```python
df.drop(columns=['Unnamed: 22', 'index'], inplace=True, errors='ignore')
print("\nAfter dropping junk cols: ", df.shape)
```

```python
df.columns = (df.columns 
              .str.strip()
              .str.lower()
              .str.replace(' ', '_')
              .str.replace('-', '_'))
print("\nFixed columns: ", df.columns.tolist())
```

```python
df['date'] = pd.to_datetime(df['date'], format='%m-%d-%y', errors='coerce')
df['amount'] = pd.to_numeric(df['amount'], errors='coerce')
df['qty'] = pd.to_numeric(df['qty'], errors='coerce')
print("\nDate sample:", df['date'].iloc[0])
print("Amount dtype:",df['amount'].dtype)
print("qty dtype:",df['qty'].dtype)
```

```python
print("\n Missing before dropna:",df.isnull().sum())
df.dropna(subset=['amount', 'date'], inplace=True)
print("Rows after dropna:", len(df))
```

```python
print("\nDuplicates found: ", df.duplicated().sum())
df.drop_duplicates(inplace=True)
print("\nDuplicates after:", df.duplicated().sum())
```

```python
df['ship_satate'] = df['ship_state'].str.strip().str.upper()
df['ship_city'] = df['ship_city'].str.strip().str.upper()
```

```python
df['revenue'] = df['amount'] * df['qty']
df['month'] = df['date'].dt.to_period('M').astype(str)
df['week'] = df['date'].dt.isocalendar().week
```

```python
df.to_csv('amazon_cleaned.csv', index=False)
print("\nFinal shape: ", df.shape)
print("\nSaved as amazon_cleaned.csv")
```
